## 1. Model Development (10 pts Required)


### 1.1 Preprocessing Steps

In this project I first loaded and merged the four provided files (final_posts.json, emotion.csv, data_identification.csv, and samplesubmission.csv) into a single dataframe. The final_posts.json file contains the raw social media posts and their hashtags (I treated hashtags as part of the text because they often contain emotional keywords (e.g., #sad, #angry, #sohappy) that directly influence classification), while emotion.csv provides the emotion labels for the training set and data_identification.csv specifies whether each example belongs to the train or test split. After merging, I split the data into df_train (labeled) and df_test (unlabeled) based on the split column.

For each post, I combined the main text with its hashtags to form a single text_all field. This is because hashtags often contain strong emotional indicators, so including them can help the classifier. I then applied a custom cleaning function called enhanced_clean_text. This function converts text to lowercase, removes URLs and user mentions, and keeps only letters, digits, whitespace, and a small set of punctuation marks (!, ?, and '). These punctuation marks are preserved because they can carry emotional emphasis (for example, “what?!” or “so happy!!!”). Finally, I normalized whitespace by collapsing multiple spaces into a single space and trimming leading and trailing spaces. The result is stored in a new clean_text column for both training and test data.

To prepare for supervised learning, I extracted clean_text as features (X_all) and the emotion column as labels (y_all) from the training subset. I then used a stratified 80/20 train–validation split to maintain the original label distribution in both sets. Since I used two different models (a LinearSVC and a DistilBERT classifier), I encoded the emotion labels into integers with LabelEncoder so that both branches could use the same label space consistently.


### 1.2 Feature Engineering Steps

For feature engineering I built a hybrid representation that combines sparse TF-IDF features with dense transformer-based embeddings.

On the classical side, I used a FeatureUnion to combine word-level and character-level TF-IDF. The word-level TfidfVectorizer uses n-grams from unigrams to trigrams, with a maximum of 80,000 features. I set min_df=2 and max_df=0.95 to filter out extremely rare and extremely common terms, and used sublinear_tf=True and English stopword removal to stabilize the term frequencies. The character-level vectorizer uses character n-grams of length 2–6, with similar min_df, max_df, and sublinear_tf settings. Character n-grams are particularly helpful for noisy social media text, where spelling variations, repeated characters, and informal patterns often matter more than vocabulary.

After fitting this TF-IDF union on the full training clean_text, I transformed the train, validation, and test sets into high-dimensional sparse vectors. I then applied an L2 Normalizer so that each example has unit norm. This step helps the LinearSVC behave more stably, since the magnitude of each feature vector is controlled.

On the neural side, I used distilbert-base-uncased as a transformer backbone. Instead of manually designing features, DistilBERT learns contextual embeddings for each token and aggregates them through a classification head into logits for the six emotion classes. I tokenized each post with a maximum sequence length of 128 tokens, with padding and truncation, and stored these as PyTorch tensors inside a custom EmotionDataset. This dataset is then fed into DataLoader objects for training and validation.


### 1.3 Explanation of Your Model

My final model is a **weighted ensemble** of two components:

* A **LinearSVC** trained on normalized TF-IDF features from the clean_text field.

* A **DistilBERT-based sequence classifier** fine-tuned on the same clean_text using class-weighted cross-entropy.

The LinearSVC branch is good at capturing surface-level lexicon and stylistic features, especially character-level patterns that show up in social media (elongated words, emotive punctuation, abbreviations, etc.). The DistilBERT branch is better at understanding contextual meaning and long-range dependencies in sentences. Both models output logits over the same six emotion classes.

During validation, I converted the SVC logits and the BERT logits into probability distributions via a softmax function. Then I formed a weighted ensemble:

* **70% weight** for DistilBERT probabilities

* **30% weight** for SVC probabilities

This reflects the fact that DistilBERT alone achieved a higher macro F1 on the validation set (around 0.52), while SVC still contributed useful complementary information. The ensemble probabilities are then used to pick the final predicted class by taking the argmax over the weighted probabilities. This weighted ensemble slightly improves the macro F1 compared to either model alone and led to my best Kaggle score (0.6645 on the leaderboard).

### 2.2 Mention Insights You Gained

A few key insights I gained from this project:

* **Text cleaning is a trade-off**. Removing too much can hurt the model, because informal punctuation and repeated characters actually carry emotional information. Keeping some of these patterns is beneficial.

* **Character n-grams are powerful for social media text**. They help the model handle spelling variations, slang, and emojis or pseudo-emojis without needing a predefined vocabulary.

* **Class imbalance really matters for macro F1**. Without class weighting, the model tends to focus too much on frequent classes like joy, and macro F1 drops because rare classes are ignored.

* **Transformers and classical models complement each other**. DistilBERT is strong on semantics, but SVC still catches certain lexical patterns that BERT may overlook. Combining the two gave me more robust predictions.

* **Weighted ensembles can be simple but effective**. I did not need a complex meta-model; just a softmax and a weighted sum of probabilities already improved performance in a stable way.

## 1. Preprocessing Steps

```python
### Add the code related to the preprocessing steps in cells inside this section

# This Python 3 environment comes with many helpful analytics libraries installed
# It is defined by the kaggle/python Docker image: https://github.com/kaggle/docker-python

import numpy as np  # linear algebra
import pandas as pd  # data processing, CSV file I/O
import os

# Inspect input directory
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))

# ============================================================
# DM Lab 2 – TF-IDF SVC + DistilBERT + Weighted Ensemble
# ============================================================

# ============================================================
# 0. BASIC IMPORTS & SEEDING
# ============================================================


import os
import json
import re
import random
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, classification_report
from sklearn.preprocessing import LabelEncoder, Normalizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import FeatureUnion
from sklearn.svm import LinearSVC

import warnings
warnings.filterwarnings("ignore")

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

# ============================================================
# 1. LOAD DATA
# ============================================================
DATA_DIR = "/kaggle/input/dm-lab-2-private-competition"

# Inspect files (optional)
for dirname, _, filenames in os.walk(DATA_DIR):
    for filename in filenames:
        print(os.path.join(dirname, filename))

with open(os.path.join(DATA_DIR, "final_posts.json"), "r", encoding="utf-8") as f:
    raw = json.load(f)

posts = pd.DataFrame([
    {
        "id": item["root"]["_source"]["post"]["post_id"],
        "text": item["root"]["_source"]["post"]["text"],
        "hashtags": item["root"]["_source"]["post"]["hashtags"],
    }
    for item in raw
])

emotion    = pd.read_csv(os.path.join(DATA_DIR, "emotion.csv"))
data_id    = pd.read_csv(os.path.join(DATA_DIR, "data_identification.csv"))
sample_sub = pd.read_csv(os.path.join(DATA_DIR, "samplesubmission.csv"))

df = posts.merge(data_id, on="id", how="inner")
df = df.merge(emotion, on="id", how="left")  # NaN emotion for test rows

df_train = df[df["split"] == "train"].copy()
df_test  = df[df["split"] == "test"].copy()

print(f"\nTrain: {len(df_train)}, Test: {len(df_test)}")
print("emotion")
print(df_train["emotion"].value_counts())


# ============================================================
# 2. TEXT PREPROCESSING
# ============================================================
def hashtags_to_str(hlist):
    if isinstance(hlist, list):
        return " ".join(hlist)
    return ""

df_train["text_all"] = df_train["text"].astype(str) + " " + df_train["hashtags"].apply(hashtags_to_str)
df_test["text_all"]  = df_test["text"].astype(str)  + " " + df_test["hashtags"].apply(hashtags_to_str)

URL_PATTERN     = r'https?://\S+|www\.\S+'
MENTION_PATTERN = r'@\w+'

def enhanced_clean_text(s):
    if pd.isna(s):
        return ""
    s = str(s).lower()
    # remove urls / mentions
    s = re.sub(URL_PATTERN, " ", s)
    s = re.sub(MENTION_PATTERN, " ", s)
    # keep letters, digits, spaces, and ! ?
    s = re.sub(r"[^a-z0-9!?'\s]", " ", s)
    # collapse whitespace
    s = re.sub(r"\s+", " ", s).strip()
    return s

df_train["clean_text"] = df_train["text_all"].apply(enhanced_clean_text)
df_test["clean_text"]  = df_test["text_all"].apply(enhanced_clean_text)


# ============================================================
# 3. TRAIN/VAL SPLIT + LABEL ENCODING
# ============================================================
X_all = df_train["clean_text"].values
y_all = df_train["emotion"].values

X_train, X_val, y_train, y_val = train_test_split(
    X_all, y_all,
    test_size=0.20,
    random_state=SEED,
    stratify=y_all
)

print("\nTrain/val sizes:", len(X_train), len(X_val))

# label encode emotions (shared by SVC & BERT & ensemble)
le = LabelEncoder()
le.fit(df_train["emotion"])
y_train_enc = le.transform(y_train)
y_val_enc   = le.transform(y_val)
num_labels  = len(le.classes_)

print("\nLabel classes:", le.classes_)
```

## 2. Feature Engineering Steps

```python
### Add the code related to the feature engineering steps in cells inside this section

# ============================================================
# 4. Feature Engineering Steps TF-IDF + LINEARSVC (SVC BRANCH)
# ============================================================

from sklearn.preprocessing import Normalizer

def build_tfidf_union(
    word_ngram=(1,3),
    char_ngram=(2,6),
    max_features=80000
):
    word_tfidf = TfidfVectorizer(
        max_features=max_features,
        ngram_range=word_ngram,
        min_df=2,
        max_df=0.95,
        sublinear_tf=True,
        stop_words='english'
    )
    char_tfidf = TfidfVectorizer(
        analyzer="char",
        ngram_range=char_ngram,
        min_df=3,
        max_df=0.95,
        sublinear_tf=True
    )
    return FeatureUnion([
        ("word", word_tfidf),
        ("char", char_tfidf),
    ])

print("\nFitting TF-IDF on full training clean_text...")
tfidf_union = build_tfidf_union(
    word_ngram=(1,3),
    char_ngram=(2,6),
    max_features=80000
)
tfidf_union.fit(df_train["clean_text"])

normalizer = Normalizer(norm="l2")

X_train_vec = tfidf_union.transform(X_train)
X_val_vec   = tfidf_union.transform(X_val)
X_test_vec  = tfidf_union.transform(df_test["clean_text"].values)

X_train_vec = normalizer.fit_transform(X_train_vec)
X_val_vec   = normalizer.transform(X_val_vec)
X_test_vec  = normalizer.transform(X_test_vec)

print("TF-IDF shapes:")
print("  X_train_vec:", X_train_vec.shape)
print("  X_val_vec:  ", X_val_vec.shape)
print("  X_test_vec: ", X_test_vec.shape)

def softmax(x, axis=1):
    # x: numpy array
    x = x - np.max(x, axis=axis, keepdims=True)
    e = np.exp(x)
    return e / np.sum(e, axis=axis, keepdims=True)

```

## 3. Model Implementation Steps

```python
### Add the code related to the model implementation steps in cells inside this section

# ============================================================
# 5. LinearSVC Training and Validation
# ============================================================

# Train SVC with best C from earlier search
svc = LinearSVC(C=0.15, class_weight="balanced", max_iter=3000)
svc.fit(X_train_vec, y_train_enc)

# get predictions & decision_function for logits
y_val_pred_svc = svc.predict(X_val_vec)
svc_val_logits = svc.decision_function(X_val_vec)  

f1_svc = f1_score(y_val_enc, y_val_pred_svc, average="macro")
print("\nSVC val macro F1:", f1_svc)
print("\nSVC validation report:\n")
print(
    classification_report(
        le.inverse_transform(y_val_enc),
        le.inverse_transform(y_val_pred_svc)
    )
)

svc_val_probs   = softmax(svc_val_logits, axis=1)
svc_test_logits = svc.decision_function(X_test_vec)
svc_test_probs  = softmax(svc_test_logits, axis=1)

# ============================================================
# 6. INSTALL TRANSFORMERS & SETUP BERT
# ============================================================

print("\nInstalling transformers (quietly)...")
!pip install -q transformers accelerate

import os as _os
_os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
_os.environ["TRANSFORMERS_NO_TF"] = "1"
_os.environ["TOKENIZERS_PARALLELISM"] = "false"

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from sklearn.utils.class_weight import compute_class_weight

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    get_linear_schedule_with_warmup,
)

# Re-seed for torch
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("\nUsing device:", device)

# ============================================================
# 7. BERT DATASETS & LOADERS
# ============================================================

model_name = "distilbert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)

MAX_LEN    = 128
BATCH_SIZE = 16  
EPOCHS     = 3
LR         = 3e-5
WEIGHT_DECAY = 0.01
WARMUP_RATIO = 0.1

class EmotionDataset(Dataset):
    def __init__(self, texts, labels=None):
        self.texts  = list(texts)
        self.labels = labels
        
    def __len__(self):
        return len(self.texts)
    
    def __getitem__(self, idx):
        text = str(self.texts[idx])
        enc = tokenizer(
            text,
            truncation=True,
            padding="max_length",
            max_length=MAX_LEN,
            return_tensors="pt"
        )
        item = {k: v.squeeze(0) for k, v in enc.items()}
        if self.labels is not None:
            item["labels"] = torch.tensor(self.labels[idx], dtype=torch.long)
        return item

train_ds = EmotionDataset(X_train, y_train_enc)
val_ds   = EmotionDataset(X_val,   y_val_enc)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False)

# ============================================================
# 8. BERT MODEL, CLASS WEIGHTS, OPTIMIZER, SCHEDULER, AMP
# ============================================================

raw_class_weights = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(y_train_enc),
    y=y_train_enc
)
raw_class_weights = np.array(raw_class_weights, dtype=np.float32)
scaled_class_weights = np.sqrt(raw_class_weights)
class_weights_tensor = torch.tensor(scaled_class_weights, dtype=torch.float).to(device)

print("\nRaw class weights:    ", raw_class_weights)
print("Scaled class weights: ", scaled_class_weights)

model = AutoModelForSequenceClassification.from_pretrained(
    model_name,
    num_labels=num_labels
)
model.to(device)

if hasattr(model, "gradient_checkpointing_enable"):
    model.gradient_checkpointing_enable()

optimizer = AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)

total_steps  = len(train_loader) * EPOCHS
warmup_steps = int(WARMUP_RATIO * total_steps)

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=warmup_steps,
    num_training_steps=total_steps
)

criterion = nn.CrossEntropyLoss(weight=class_weights_tensor)

from torch.cuda.amp import autocast, GradScaler
use_amp = (device.type == "cuda")
scaler = GradScaler(enabled=use_amp)

# ============================================================
# 9. TRAINING + EVAL FUNCTIONS
# ============================================================

def eval_model(model, loader):
    model.eval()
    all_preds  = []
    all_labels = []
    all_logits = []
    with torch.no_grad():
        for batch in loader:
            input_ids      = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            labels         = batch["labels"].to(device)

            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )
            logits = outputs.logits
            preds  = torch.argmax(logits, dim=1)

            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
            all_logits.append(logits.cpu().numpy())
    all_preds  = np.array(all_preds)
    all_labels = np.array(all_labels)
    all_logits = np.concatenate(all_logits, axis=0)
    macro_f1 = f1_score(all_labels, all_preds, average="macro")
    return macro_f1, all_preds, all_labels, all_logits

best_f1 = 0.0
patience = 1
no_improve = 0

print("\nStarting DistilBERT training...\n")
for epoch in range(EPOCHS):
    model.train()
    total_loss = 0.0

    for batch in train_loader:
        optimizer.zero_grad()

        input_ids      = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        labels         = batch["labels"].to(device)

        with autocast(enabled=use_amp):
            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )
            logits = outputs.logits
            loss = criterion(logits, labels)

        if use_amp:
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
        else:
            loss.backward()
            optimizer.step()

        scheduler.step()
        total_loss += loss.item()
    
    avg_loss = total_loss / len(train_loader)
    val_f1, val_preds_enc_bert, val_true_enc_bert, val_logits_bert = eval_model(model, val_loader)

    print(f"Epoch {epoch+1}/{EPOCHS} - train loss {avg_loss:.4f} - val F1 {val_f1:.4f}")
    
    if val_f1 >= best_f1:
        best_f1 = val_f1
        no_improve = 0
        torch.save(model.state_dict(), "best_distilbert_full.pt")
        print("  -> New best BERT model saved")
        # store last best val logits / preds for ensemble
        best_val_logits_bert = val_logits_bert
        best_val_preds_bert  = val_preds_enc_bert
        best_val_true_bert   = val_true_enc_bert
    else:
        no_improve += 1
        if no_improve > patience:
            print("Early stopping triggered.")
            break

# Reload best model
best_bert = AutoModelForSequenceClassification.from_pretrained(
    model_name,
    num_labels=num_labels
)
best_bert.load_state_dict(torch.load("best_distilbert_full.pt", map_location=device))
best_bert.to(device)
best_bert.eval()

# ============================================================
# 10. BERT VALIDATION METRICS
# ============================================================

bert_val_f1, bert_val_preds_enc, bert_val_true_enc, bert_val_logits = eval_model(best_bert, val_loader)
bert_val_probs = softmax(bert_val_logits, axis=1)

bert_val_pred_labels = le.inverse_transform(bert_val_preds_enc)
bert_val_true_labels = le.inverse_transform(bert_val_true_enc)

print("\nDistilBERT full fine-tune val macro F1:", bert_val_f1)
print("\nBERT validation classification report:\n")
print(classification_report(bert_val_true_labels, bert_val_pred_labels))


# ============================================================
# 11. WEIGHTED ENSEMBLE ON VALIDATION
# ============================================================

ALPHA_BERT = 0.7   # weight for BERT, 0.3 for SVC

assert svc_val_probs.shape == bert_val_probs.shape

ensemble_val_probs = ALPHA_BERT * bert_val_probs + (1.0 - ALPHA_BERT) * svc_val_probs
ensemble_val_preds_enc = np.argmax(ensemble_val_probs, axis=1)

f1_ensemble = f1_score(y_val_enc, ensemble_val_preds_enc, average="macro")
ensemble_val_pred_labels = le.inverse_transform(ensemble_val_preds_enc)
ensemble_val_true_labels = le.inverse_transform(y_val_enc)

print("\n====================")
print("VALIDATION SUMMARY")
print("====================")
print(f"SVC macro F1:       {f1_svc:.4f}")
print(f"BERT macro F1:      {bert_val_f1:.4f}")
print(f"Ensemble macro F1:  {f1_ensemble:.4f}")

print("\nEnsemble validation classification report:\n")
print(classification_report(ensemble_val_true_labels, ensemble_val_pred_labels))

# ============================================================
# 12. RUN BEST BERT ON TEST SET
# ============================================================

test_ds = EmotionDataset(df_test["clean_text"].tolist(), labels=None)
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False)

bert_test_logits = []
best_bert.eval()
with torch.no_grad():
    for batch in test_loader:
        input_ids      = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)

        outputs = best_bert(
            input_ids=input_ids,
            attention_mask=attention_mask
        )
        logits = outputs.logits
        bert_test_logits.append(logits.cpu().numpy())

bert_test_logits = np.concatenate(bert_test_logits, axis=0)
bert_test_probs  = softmax(bert_test_logits, axis=1)

# ============================================================
# 13. ENSEMBLE ON TEST + SUBMISSION
# ============================================================

ensemble_test_probs = ALPHA_BERT * bert_test_probs + (1.0 - ALPHA_BERT) * svc_test_probs
ensemble_test_preds_enc = np.argmax(ensemble_test_probs, axis=1)
ensemble_test_pred_labels = le.inverse_transform(ensemble_test_preds_enc)

submission = pd.DataFrame({
    "id": df_test["id"].values,
    "emotion": ensemble_test_pred_labels
})

submission.to_csv("submission.csv", index=False)
print("\nSaved submission.csv (Ensemble: 0.7*BERT + 0.3*SVC)")

```
